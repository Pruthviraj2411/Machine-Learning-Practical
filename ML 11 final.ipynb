```python
import gymnasium as gym
import numpy as np
```

```python
# Create Taxi environment
env = gym.make("Taxi-v4", render_mode="ansi")

# Create Q-table
q_table = np.zeros((env.observation_space.n, env.action_space.n))
```

```python
# Parameters
learning_rate = 0.1
discount_factor = 0.9
epsilon = 0.1
episodes = 5000
```

```python
# Train the taxi agent
for episode in range(episodes):

    state, info = env.reset()
    done = False

    while not done:
        # Choose an action
        if np.random.random() < epsilon:
            action = env.action_space.sample()
        else:
            action = np.argmax(q_table[state])

        # Perform the action
        next_state, reward, terminated, truncated, info = env.step(action)
        done = terminated or truncated

        # Update Q-table
        q_table[state, action] += learning_rate * (reward + discount_factor * np.max(q_table[next_state])- q_table[state, action])

        state = next_state
```

```python
# Test the trained taxi agent
state, info = env.reset()
done = False
total_reward = 0

print("Taxi Environment:")

while not done:
    action = np.argmax(q_table[state])

    state, reward, terminated, truncated, info = env.step(action)
    done = terminated or truncated

    total_reward += reward

    print(env.render())

print("Total Reward:", total_reward)

env.close()
```

Let's understand your Taxi Problem Using Reinforcement Learning code line by line in simple words for your practical exam.

1. Import libraries
import gymnasium as gym
import numpy as np
gymnasium: Provides the Taxi environment.
numpy: Helps create and manage the Q-table.
2. Create the environment
env = gym.make("Taxi-v3")

Creates the Taxi game where the taxi must pick up a passenger and drop them at the correct destination.

3. Create the Q-table
q = np.zeros((env.observation_space.n, env.action_space.n))

Creates a Q-table filled with zeros.

Rows: Possible states of the taxi.
Columns: Possible actions.
Q-values: Indicate how useful an action is in a particular state.

The taxi learns by updating these values.

4. Start training
for episode in range(1000):
    state, _ = env.reset()
    done = False
range(1000): Trains the taxi for 1,000 episodes.
env.reset(): Starts a new episode.
state: Represents the current situation of the taxi.
done = False: Indicates that the task is not finished.
5. Choose an action
while not done:
    action = env.action_space.sample() if np.random.random() < 0.1 else np.argmax(q[state])

The taxi chooses an action in two ways:

10% of the time: Chooses a random action to explore.
90% of the time: Chooses the action with the highest Q-value.

This helps the taxi explore new actions and use what it has learned.

6. Perform the action
next_state, reward, terminated, truncated, _ = env.step(action)
done = terminated or truncated
env.step(action): Performs the selected action.
next_state: The new state after the action.
reward: The reward or penalty received.
terminated: Indicates whether the task has finished.
truncated: Indicates whether the episode ended because of a time limit.
done: Becomes True when the episode ends.
7. Update the Q-table
q[state, action] += 0.1 * (reward + 0.9 * np.max(q[next_state]) - q[state, action])

This is the main Q-learning formula.

0.1: Learning rate — how quickly the taxi learns.
reward: Reward received for the action.
0.9: Discount factor — importance given to future rewards.
np.max(q[next_state]): Highest Q-value in the next state.
q[state, action]: Current Q-value being updated.

In simple words: The taxi updates its knowledge based on the reward it receives and the future rewards it expects.

state = next_state

Updates the current state to the new state so the taxi can continue learning.

8. Test the trained taxi
state, _ = env.reset()
done = False

Starts a new episode to test what the taxi has learned.

while not done:
    action = np.argmax(q[state])
    state, reward, terminated, truncated, _ = env.step(action)
    done = terminated or truncated
Selects the best-known action using np.argmax(q[state]).
Performs the action.
Updates the state.
Continues until the episode ends.

During testing, the taxi chooses its best-known action rather than exploring randomly.

9. Display the result
print("Taxi completed the task!")
env.close()
print(): Displays a completion message.
env.close(): Closes the environment.

Note: The message is printed when the episode ends, even if the taxi did not successfully drop off the passenger. To confirm successful completion, you would need to check the final reward or task status.
